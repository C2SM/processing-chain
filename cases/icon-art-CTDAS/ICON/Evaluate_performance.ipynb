```python
import xarray as xr
import matplotlib.pyplot as plt
from unidecode import unidecode
import numpy as np
import datetime as dt
import pandas as pd

# Station dictionary with measurement uncertainty
mdm_dictionary = {
    'Beromunster_212': 8.2383423, 'Bilsdale_248': 5.8534036, 'Biscarrosse_47': 5.5997221,
    'Cabauw_207': 8.6093283, 'Carnsore Point_14': 4.1894007, 'Ersa_40': 4.3997285,
    'Gartow_341': 6.570544, 'Heidelberg_30': 10.660628, 'Hohenpeissenberg_131': 6.0553513,
    'Hyltemossa_150': 5.485432, 'Ispra_100': 11.612817, 'Jungfraujoch_13': 3.0802848,
    'Karlsruhe_200': 10.05013, 'Kresin u Pacova_250': 5.829324, 'Heathfield_100': 6.6675706,
    'La Muela_80': 5.2093291, 'Laegern-Hochwacht_32': 11.556924, 'Lindenberg_98': 7.4387555,
    'Lutjewad_60': 7.651525, 'Monte Cimone_8': 3.7325112,
    'Observatoire de Haute Provence_100': 6.146905,
    "Observatoire perenne de l'environnement_120": 8.8854113, 'Pic du Midi_28': 3.2196398,
    'Plateau Rosa_10': 3.3211231, 'Puy de Dome_10': 5.4529948, 'Ridge Hill_90': 7.0861707,
    'Saclay_100': 8.8669567, 'Schauinsland_12': 5.7896755, 'Tacolneston_185': 6.6675706,
    'Torfhaus_147': 6.622525, 'Trainou_180': 7.821612, 'Weybourne_10': 6.4674397,
    'Zugspitze_3': 3.6796716,
}

rmse_post = {}

# Stations using the midnight-to-morning window (00:00 - 07:00)
nighttime_stations = {
    'Jungfraujoch_13', 'Monte Cimone_8', 'Puy de Dome_10', 'Pic du Midi_28', 'Zugspitze_3',
    'Hohenpeissenberg_131', 'Schauinsland_12', 'Plateau Rosa_10'
}

resample = True
mdm_dict = {}
station_times_dict = {}

for selected_station in mdm_dictionary.keys():
    print(f"Processing station: {selected_station}")

    # Reset start and end date for each station
    startdate = dt.datetime(2018, 1, 1)
    enddate = dt.datetime(2018, 12, 17)

    # Reset data storage for each station
    ICON_values = []
    ICOS_values = []
    post_values = []
    ICON_dates = []
    ICOS_dates = []

    # Identify which time window to use
    if selected_station in nighttime_stations:
        time_start = 0  # 00:00
        time_end = 7  # 07:00
    else:
        time_start = 12  # 12:00
        time_end = 17  # 17:00

    while startdate < enddate:
        try:
            ICON = xr.open_dataset(f"/capstor/scratch/cscs/ekoene/extracted_ICOS/runthrough_{startdate.strftime('%Y%m%d')}.nc")
            ICONopt = xr.open_dataset(f"/capstor/scratch/cscs/ekoene/processing-chain/work/icon-art-CTDAS/global_outputs/extracted_ICOS/opt2_{startdate.strftime('%Y%m%d')}.nc")
            ICOS = xr.open_dataset(f"/capstor/scratch/cscs/ekoene/processing-chain/work/icon-art-CTDAS/global_inputs/ICOS/Extracted_{startdate.strftime('%Y%m%d')}_{(startdate+dt.timedelta(days=11)).strftime('%Y%m%d')}_alldates_masl.nc")

            full = (ICON.TRCO2_A + ICON.TRCO2_BG + ICON.CO2_RA - ICON.CO2_GPP + ICON.biosource - ICON.biosink) / (1 - ICON.qv) * 28.97 / 44.01 * 1e6
            post = (ICONopt["TRCO2_A_ENS"][0,:,:].squeeze() + ICONopt.biosource - ICONopt.biosink) / (1 - ICONopt.qv) * 28.97 / 44.01 * 1e6
            
            stations = np.array([unidecode(x) for x in ICON.site_name.values])
            stations_post = np.array([unidecode(x) for x in ICONopt.site_name.values])
            mdm_keys = np.array([unidecode(k) for k in mdm_dictionary.keys()])

            # Ensure the station exists
            if selected_station not in mdm_keys:
                raise ValueError(f"Station '{selected_station}' not found in mdm_dictionary")

            # Get the station index
            station_index = np.where(stations == selected_station)[0]
            if len(station_index) == 0:
                print(f"Skipping {selected_station}: Not found in ICON dataset on {startdate}")
                startdate += dt.timedelta(days=10)
                continue
            station_index = station_index[0]

            # Get the station index
            station_index_post = np.where(stations_post == selected_station)[0]
            if len(station_index_post) == 0:
                print(f"Skipping {selected_station}: Not found in ICON dataset on {startdate}")
                startdate += dt.timedelta(days=10)
                continue
            station_index_post = station_index_post[0]

            # Filter by time range
            # Ensure time is in datetime64 format
            ICON_times = ICON.time.values.astype('datetime64[h]')[24:]  # Convert to hourly datetime

            # Extract the hours using numpy
            time_hours = np.array([t.astype(object).hour for t in ICON_times])            
            time_mask = (time_hours >= time_start) & (time_hours < time_end)

            full['time'] = pd.DatetimeIndex(full['time'].values)
            post['time'] = pd.DatetimeIndex(post['time'].values)
            ICOS["time"] = pd.DatetimeIndex(ICOS.Dates[0].values)
            full = full.isel(sites=station_index)[24:]
            post = post.isel(sites=station_index_post)[24:]
            ICOS = ICOS.Concentration.isel(station=station_index)
            if selected_station in nighttime_stations:
                full = full.sel(time=full.time.dt.hour.isin([0, 1, 2, 3, 4, 5, 6, 7]))
                post = post.sel(time=post.time.dt.hour.isin([0, 1, 2, 3, 4, 5, 6, 7]))
                ICOS = ICOS.sel(time=ICOS.time.dt.hour.isin([0, 1, 2, 3, 4, 5, 6, 7]))
            else:
                full = full.sel(time=full.time.dt.hour.isin([12, 13, 14, 15, 16]))
                post = post.sel(time=post.time.dt.hour.isin([12, 13, 14, 15, 16]))
                ICOS = ICOS.sel(time=ICOS.time.dt.hour.isin([12, 13, 14, 15, 16]))
            
            if resample:
                full = full.resample(time='D').mean("time")
                post = post.resample(time='D').mean("time")
                ICOS = ICOS.resample(time='D').mean("time")

            ICON_values.append(full)
            post_values.append(post)
            ICOS_values.append(ICOS)
            ICON_dates.append(full.time)
            ICOS_dates.append(ICOS.time)

        except FileNotFoundError as e:
            print(f"Skipping date {startdate} due to missing file: {e}")

        except Exception as e:
            print(f"Error on {startdate}: {e}")

        startdate += dt.timedelta(days=10)

    # Flatten arrays
    ICON_values = np.asarray(ICON_values).flatten()
    post_values = np.asarray(post_values).flatten()
    ICOS_values = np.asarray(ICOS_values).flatten()
    ICON_dates = np.asarray(ICON_dates).flatten()
    ICOS_dates = np.asarray(ICOS_dates).flatten()

    if len(ICON_values) == 0 or len(ICOS_values) == 0:
        print(f"No data available for station {selected_station}")
        continue

    # Align datasets by timestamps
    common_dates, icon_idx, icos_idx = np.intersect1d(ICON_dates, ICOS_dates, return_indices=True)

    icon_common = ICON_values[icon_idx]
    post_common = post_values[icon_idx]
    icos_common = ICOS_values[icos_idx]

    # Remove NaN values
    valid_mask = ~np.isnan(icos_common)
    icon_valid = icon_common[valid_mask]
    post_valid = post_common[valid_mask]
    icos_valid = icos_common[valid_mask]

    # Calculate RMSE
    mdm_dict[selected_station] = post_valid - icos_valid
    station_times_dict[selected_station] = ICON_dates[icon_idx][valid_mask]

def compute_correlation(x, y):
    """
    Manually compute the Pearson correlation coefficient between two variables x and y.
    """
    # Mean of x and y
    mean_x = np.mean(x)
    mean_y = np.mean(y)

    # Compute the numerator and the denominator of the Pearson correlation formula
    numerator = np.sum((x - mean_x) * (y - mean_y))
    denominator = np.sqrt(np.sum((x - mean_x) ** 2) * np.sum((y - mean_y) ** 2))

    # Return the correlation coefficient
    return numerator / denominator

# Create a dictionary to store correlation values for all pairs of stations
correlation_results = {}

# Iterate over all pairs of stations
for station_1 in mdm_dictionary:
    print(f"Processing station: {station_1} for the MDM correlations")
    for station_2 in mdm_dictionary:
        # Get the overlapping times for station_1 and station_2
        common_times = set(station_times_dict[station_1]) & set(station_times_dict[station_2])
        
        if common_times:  # If there is any overlap
            # Get the MDM values for the common times
            mdm_1 = [mdm_dict[station_1][np.where(station_times_dict[station_1] == time)[0][0]] for time in common_times]
            mdm_2 = [mdm_dict[station_2][np.where(station_times_dict[station_2] == time)[0][0]] for time in common_times]
            
            # Compute the correlation between the MDM values for this pair of stations
            correlation = compute_correlation(np.array(mdm_1), np.array(mdm_2))
            correlation_results[(station_1, station_2)] = correlation
        else:
            correlation_results[(station_1, station_2)] = np.nan

station_names = list(mdm_dictionary.keys())
correlation_matrix = np.zeros((len(station_names), len(station_names)))

# Fill the correlation matrix with the computed values
for i, station_1 in enumerate(station_names):
    for j, station_2 in enumerate(station_names):
        if station_1 != station_2:
            # Get the correlation for the pair (station_1, station_2)
            correlation = correlation_results.get((station_1, station_2), None)
            if correlation is not None:
                correlation_matrix[i, j] = correlation
            else:
                # If there's no correlation (no overlap), fill with NaN
                correlation_matrix[i, j] = np.nan
        else:
            # Set the diagonal to 1 (100% correlation with itself)
            correlation_matrix[i, j] = 1
```
